```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from bs4 import BeautifulSoup

driver = webdriver.Chrome()

driver.get("https://www.icy-veins.com/d2/terror-zones-guide")

page_source = driver.page_source

soup = BeautifulSoup(page_source, 'html.parser')

driver.close()
```

```python
span = soup.find_all('li')
zones = []

for i in span[21:57]:
    zone = i.text.replace("and", "").replace("/", "").replace(",", "").replace("  ", " ")
    
    zones.append(zone)
    
print(zones)
    
```

```python
# Extract the raw HTML content inside the span
raw_html = span.decode_contents()

# Split the content by <br/> tags
text_parts = raw_html.split('<br/>')

# Remove any surrounding whitespace
text_parts = [text.strip() for text in text_parts]

text_parts
```
